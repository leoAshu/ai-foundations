## Indexing & Slicing

```python
X = torch.randn((3, 4))
X
```

```python
X[0], X[1: 3], X[-1]
```

```python
X[1:3, :2] = -0.9999
X
```

## Operations

```python
torch.exp(X)
```

```python
X = torch.tensor([[1, 2, 3], [4, 5, 6]])
Y = torch.tensor([[4, 5, 6], [1, 2, 3]])

X + Y, X - Y, X * Y, X / Y, X ** Y
```

```python
torch.cat((X, Y), dim=0), torch.cat((X, Y), dim=1)
```

```python
X == Y
```

```python
torch.sum(X), torch.sum(X, 0), torch.sum(X, 1), torch.sum(X, 0, keepdim=True), torch.sum(X, 1, keepdim=True)
```

## Broadcasting

```python
x = torch.arange(2).reshape((1, 2))
y = torch.arange(3).reshape((3, 1))

x, y
```

```python
x + y, x * y
```

## Saving Memory

```python
X, Y
```

```python
before = id(Y)
Y = X + Y

before == id(Y)
```

```python
before = id(Y)
Y[:] = X + Y

before == id(Y)
```

## Type Conversion

```python
A = X.numpy()
B = torch.from_numpy(A)

X, A, B
```

```python
B[0, 0] = 0

X, A, B
```

```python
type(X), type(A), type(B)
```

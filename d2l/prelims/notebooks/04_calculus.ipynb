```python
import torch
```

```python
x = torch.arange(4, dtype=torch.float32)

x
```

```python
x.requires_grad_(True)

y = torch.dot(x, x)

y
```

```python
y.backward()

x.grad
```

```python
x.grad == 2 * x
```

```python
x.grad.zero_()

y = torch.sum(x)

y.backward()
x.grad
```

## Backward for Non-scalar

```python
x.grad.zero_()
y = x + x

y
```

```python
y.sum().backward()

x.grad
```

## Detaching Computation

```python
x
```

```python
x.grad.zero_()

y = x * x
z = y * x

z.sum().backward()

x.grad, x.grad == 3 * x**2
```

```python
x.grad.zero_()

y = x * x
u = y.detach()
z = u * x

z.sum().backward()

x.grad, x.grad == u
```

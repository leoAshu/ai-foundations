## Tensor Arithmetic

```python
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
B = A.clone()

A, A + B
```

```python
A * B
```

```python
1 + A
```

```python
A + 1
```

## Reduction

```python
torch.sum(A), torch.sum(A, axis=[0, 1])
```

```python
torch.sum(A, axis=0), torch.sum(A, axis=1)
```

```python
torch.mean(A, axis=0), torch.sum(A, axis=0) / A.shape[0]
```

```python
torch.mean(A), torch.sum(A) / torch.numel(A)
```

## Non Reduction

```python
torch.sum(A, axis=0, keepdim=True), torch.sum(A, axis=1, keepdim=True)
```

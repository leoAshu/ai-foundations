```python
import matplotlib.pyplot as plt
import torch
```

```python
x = torch.arange(-10, 10+0.2, 0.2, requires_grad=True)

x.detach()[:3]
```

## ReLU

```python
y = torch.relu(x)

y.detach()[:3]
```

```python
plt.plot(x.detach(), y.detach())
plt.grid()
```

```python
y.sum().backward()
```

```python
plt.plot(x.detach(), x.grad)
plt.grid()
```

## Sigmoid

```python
y = torch.sigmoid(x)

y.detach()[:3]
```

```python
plt.plot(x.detach(), y.detach())
plt.grid()
```

```python
x.grad.zero_()
y.sum().backward()
```

```python
plt.plot(x.detach(), x.grad)
plt.grid()
```

## Tanh

```python
y = torch.tanh(x)

y.detach()[:3]
```

```python
plt.plot(x.detach(), y.detach())
plt.grid()
```

```python
x.grad.zero_()

y.sum().backward()
```

```python
plt.plot(x.detach(), x.grad)
plt.grid()
```
